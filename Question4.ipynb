 You have students.csv (student_id, name, class) and scores.csv (student_id, subject, score): 

- (a) Read both into DataFrames and merge on student_id. 
- (b) Use groupby to find average score per subject and per student. 
- (c) Add a grade column: 'A' >= 85, 'B' >= 70, 'C' >= 55, 'F' otherwise. 
- (d) Use numpy: mean, standard deviation, 75th percentile of all scores. 
- (e) Predict the output: 
import pandas as pd 
data = {'name':['A','B','A','B'], 
'score':[80,90,70,85]} 
df = pd.DataFrame(data) 
result = df.groupby('name')['score'].mean() 
print(result.to_dict()) 

**(a) Read both into DataFrames and merge on student_id.**

```python
import pandas as pd
```

```python
students = pd.read_csv("students.csv", skipinitialspace=True)
scores = pd.read_csv("scores.csv", skipinitialspace=True)
```

```python
print(students)
print(type(scores))
```

**(b) Use groupby to find average score per subject and per student.**

```python
## In order to use the group by we need to merge the both dataframes
data = pd.merge(students, scores,on='student_id')
```

```python
data
```

```python
## Avg score per subject
subject_average = data.groupby("subject")["score"].mean()

print(subject_average)
```

```python
## Avg score per student

subject_average = data.groupby("name")["score"].mean()

print(subject_average)
```

**(c) Add a grade column: 'A' >= 85, 'B' >= 70, 'C' >= 55, 'F' otherwise.**

```python
def get_grade(score):
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 55:
        return "C"
    else:
        return "F"
```

```python
data["grade"] = data["score"].apply(get_grade)
```

```python
data
```

**(d) Use numpy: mean, standard deviation, 75th percentile of all scores.**

```python
import numpy as np
```

```python
type(data['score'])
```

```python
mean = np.mean(data['score'])
print(mean)

std = np.std(data['score'])
print(std)

percentile_75 = np.percentile(data['score'], 75)
print(percentile_75)
```
