 Write a Python program that accepts a user's name and age as string inputs. 

 Perform the following operations and print clearly labelled results: 

 (a) Convert the age to an integer and calculate the year they were born (assume current year = 2025). 

 (b) Check if the age is even or odd using the modulo operator. 

 (c) Print the age as a float divided by the life expectancy of 80, formatted to 4 decimal places.

 (d) Use type() to print the data type of each: name, age (before and after conversion). 
 
 (e) Predict the output: x = '15' y = int(x) z = float(y) / 4 print(type(x).__name__, y * 2, round(z, 2), y % 2 == 0) 


**(a) Convert the age to an integer and calculate the year they were born (assume current year = 2025).**


```python
print("Hello")
```

```python
name = input("Enter your name: ")
age = input("Enter your age: ")

age_int = int(age)

birth_year = 2025 - age_int

print("Name:", name)
print("Birth Year:", birth_year)
```

**(b) Check whether age is even or odd**

```python
if age_int % 2 == 0:
    print("Age is Even")
else:
    print("Age is Odd")
```

**(c) Age divided by life expectancy**

```python
life_expectancy = 80

ratio = age_int / life_expectancy

print("Age / Life Expectancy:", format(ratio, ".4f"))
```

**d) Print data types before and after conversion**

```python
name = input("Enter your name: ")
age = input("Enter your age: ")

print("Before Conversion:")
print("Name:", type(name))
print("Age:", type(age))

age = int(age)

print("After Conversion:")
print("Name:", type(name))
print("Age:", type(age))
```
