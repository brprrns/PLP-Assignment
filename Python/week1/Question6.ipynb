**(a) safe_divide(a, b): handle ZeroDivisionError and TypeError. Show 3 test calls.**

```python
def safe_divide(a, b):
    try:
        res = a/b
    except ZeroDivisionError as e:
        print(e)
    except TypeError as e:
        print("Type Error: ", e)
    else:
        print("result ", res)
    finally :
        print("Divison Completed")
```

```python
print(safe_divide(8,8))
print(safe_divide(8,0))
print(safe_divide(8,"Raj"))
```

**(b) read_student_file(filename): handle FileNotFoundError, use finally to print 'File operation complete'.** 

```python
def read_student_file(filename):
    try:
        with open(filename, "r") as file:
            print(file.read())
    except FileNotFoundError as e:
        print("File not found", e)
    finally :
        print("File operation complete")
```

**(c) Write a main block calling both functions with values that trigger each exception and 
values that succeed.**

```python
if __name__ == "__main__":

    # 3 test calls for safe_divide()
    safe_divide(10, 2)       # Successful
    safe_divide(10, 0)       # ZeroDivisionError
    safe_divide(10, "2")     # TypeError

    # Testing read_student_file()
    read_student_file("students.txt")        # File may exist
    read_student_file("unknown.txt")        # FileNotFoundError
```
